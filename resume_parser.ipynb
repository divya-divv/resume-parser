```python
# ==========================================
# RESUME PARSER PROJECT
# ==========================================

import re

# Sample resume text
resume_text = """
DIVYA J

Computer Science Engineering Student

Email: divya@example.com
Phone: +91 9876543210

EDUCATION
Bachelor of Engineering in Computer Science Engineering

SKILLS
Python, Java, SQL, HTML, CSS, Machine Learning,
Git, GitHub, Data Science

PROJECTS
Sentiment Analysis of Reviews
Object Detection using YOLO

CERTIFICATIONS
Python Programming
Machine Learning
"""

# ------------------------------------------
# 1. Extract Email
# ------------------------------------------

email = re.findall(
    r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b',
    resume_text
)

# ------------------------------------------
# 2. Extract Phone Number
# ------------------------------------------

phone = re.findall(
    r'(?:\+91[\s-]?)?[6-9]\d{9}',
    resume_text
)

# ------------------------------------------
# 3. Find Skills
# ------------------------------------------

skills_list = [
    "Python",
    "Java",
    "C++",
    "SQL",
    "HTML",
    "CSS",
    "JavaScript",
    "Machine Learning",
    "Data Science",
    "Git",
    "GitHub",
    "Artificial Intelligence",
    "Deep Learning"
]

found_skills = []

for skill in skills_list:
    if re.search(
        r'\b' + re.escape(skill) + r'\b',
        resume_text,
        re.IGNORECASE
    ):
        found_skills.append(skill)

# ------------------------------------------
# 4. Find Education
# ------------------------------------------

education_keywords = [
    "B.E",
    "B.Tech",
    "Bachelor",
    "Computer Science",
    "Engineering",
    "MCA",
    "M.Tech"
]

found_education = []

for education in education_keywords:
    if re.search(
        re.escape(education),
        resume_text,
        re.IGNORECASE
    ):
        found_education.append(education)

# ------------------------------------------
# 5. Display Results
# ------------------------------------------

print("=" * 50)
print("           RESUME PARSER RESULTS")
print("=" * 50)

print("\nName:")
print("DIVYA J")

print("\nEmail:")
print(email[0] if email else "Not found")

print("\nPhone:")
print(phone[0] if phone else "Not found")

print("\nSkills:")
print(", ".join(found_skills))

print("\nEducation:")
print(", ".join(found_education))

print("\nProjects:")
print("Sentiment Analysis of Reviews")
print("Object Detection using YOLO")

print("\n" + "=" * 50)
print("Resume parsing completed successfully! ✅")
print("=" * 50)
```
